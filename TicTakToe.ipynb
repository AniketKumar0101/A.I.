```python
import random

board = [" " for _ in range(9)]

def display():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")
    print()

def check_winner(player):
    wins = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for win in wins:
        if board[win[0]] == player and \
           board[win[1]] == player and \
           board[win[2]] == player:
            return True

    return False

def ai_move():
    empty = []

    for i in range(9):
        if board[i] == " ":
            empty.append(i)

    if empty:
        position = random.choice(empty)
        board[position] = "O"
        print(f"AI chooses position {position + 1}")

def draw():
    return " " not in board

print("TIC-TAC-TOE")
print("You = X")
print("AI  = O")

print()
print("Positions:")
print(" 1 | 2 | 3 ")
print("---+---+---")
print(" 4 | 5 | 6 ")
print("---+---+---")
print(" 7 | 8 | 9 ")

while True:
    display()

    position = input("Enter your position (1-9): ")

    if not position.isdigit():
        print("Enter a number between 1 and 9.")
        continue

    position = int(position)

    if position < 1 or position > 9:
        print("Invalid position.")
        continue

    position -= 1

    if board[position] != " ":
        print("Position already occupied.")
        continue

    board[position] = "X"

    if check_winner("X"):
        display()
        print("You win!")
        break

    if draw():
        display()
        print("Game Draw!")
        break

    ai_move()

    if check_winner("O"):
        display()
        print("AI wins!")
        break

    if draw():
        display()
        print("Game Draw!")
        break
```
