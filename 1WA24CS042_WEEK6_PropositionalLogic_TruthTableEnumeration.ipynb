```python
import itertools

def evaluate_sentence(sentence, model):

    return sentence(model)

def tt_check_all(kb, alpha, symbols, model):

    if not symbols:
        if evaluate_sentence(kb, model):
            return evaluate_sentence(alpha, model)
        else:
            return True

    p = symbols[0]
    rest = symbols[1:]

    model_true = model.copy()
    model_true[p] = True

    model_false = model.copy()
    model_false[p] = False

    return tt_check_all(kb, alpha, rest, model_true) and tt_check_all(kb, alpha, rest, model_false)

def tt_entails(kb, alpha, symbols):

    return tt_check_all(kb, alpha, list(symbols), {})

def print_truth_table(symbols, kb, alpha):

    print("\n" + "="*50)
    print(" TRUTH TABLE ENUMERATION ")
    print("="*50)

    header = " | ".join(symbols) + " | KB | α "
    print(header)
    print("-" * len(header))

    for combinations in itertools.product([False, True], repeat=len(symbols)):
        model = dict(zip(symbols, combinations))

        kb_val = evaluate_sentence(kb, model)
        alpha_val = evaluate_sentence(alpha, model)

        row_str = " | ".join(f"{str(model[s]):<5}" for s in symbols)
        row_str += f" | {str(kb_val):<5} | {str(alpha_val):<5}"
        print(row_str)

    print("-" * len(header))


if __name__ == "__main__":

    symbols_slide = ['A', 'B', 'C']

    kb_slide = lambda m: (m['A'] or m['C']) and (m['B'] or not m['C'])
    alpha_slide = lambda m: m['A'] or m['B']

    print("Checking Entailment for Slide Example:")
    print("KB = (A ∨ C) ∧ (B ∨ ¬C)")
    print("α  = A ∨ B")

    print_truth_table(symbols_slide, kb_slide, alpha_slide)

    entails = tt_entails(kb_slide, alpha_slide, symbols_slide)
    print(f"\nResult: Does KB ⊨ α? -> {entails}")
```
